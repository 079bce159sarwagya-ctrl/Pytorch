Tensor Operations On GPU

```python
import torch
torch.cuda.is_available()

device=torch.device('cuda')

print(torch.rand((2,3),device=device))
print(torch.rand(2,3).to(device))



```

```python
import time
#Define the size of the matrices
size = 10000

#Create random matrices on CPU
matrix_cpu1= torch.rand(size, size)
matrix_cpu2= torch.rand(size, size)

#Measure the time on CPU
start=time.time()
result=torch.matmul(matrix_cpu1, matrix_cpu2)
end=time.time()
print(result)
CPU=end-start
print(f"Time taken on CPU: {end-start} seconds")

#Change the matrices to GPU
matrix_gpu1=matrix_cpu1.to(device)
matrix_gpu2=matrix_cpu2.to(device)

#Measure the time on GPU

start=time.time()
result=torch.matmul(matrix_gpu1, matrix_gpu2)
end=time.time()
print(result)
GPU=end-start
print(f"Time taken on GPU: {end-start} seconds")

#Compare the difference in time
print(f"GPU is {CPU/GPU} times faster than CPU")

```
